```python
import pandas as pd
import pandera as pa
from pandera import Column, Check, DataFrameSchema

def infer_schema_from_df(df: pd.DataFrame) -> DataFrameSchema:
    columns_dict = {}

    for col in df.columns:
        dtype = df[col].dtype
        series = df[col]

        # Определяем тип Pandera по dtype pandas
        if pd.api.types.is_integer_dtype(dtype):
            pa_type = int
        elif pd.api.types.is_float_dtype(dtype):
            pa_type = float
        elif pd.api.types.is_string_dtype(dtype) or dtype == object:
            pa_type = str
        else:
            # fallback: оставляем как есть, Pandera попробует разобраться
            pa_type = dtype

        checks = []

        # nullable: если есть хотя бы один NaN — nullable=True, иначе False
        nullable = series.isna().any()

        # Для числовых добавляем min/max, если они осмысленны
        if pd.api.types.is_numeric_dtype(dtype) and not nullable:
            min_val = series.min()
            max_val = series.max()
            # Можно добавить небольшой «запас» на будущее, например ±1%
            # checks.append(Check.ge(min_val * 0.99))
            # checks.append(Check.le(max_val * 1.01))
            checks.append(Check.in_range(min_val, max_val))

        columns_dict[col] = Column(
            pa_type,
            nullable=nullable,
            checks=checks if checks else None,
        )

    return DataFrameSchema(columns_dict)


# Пример использования
# df = pd.read_csv("data.csv")  # ваш DataFrame

# Выводим статистику (как у вас в цикле)
for cl in df.columns:
    print("*" * 20 + cl + "*" * 20)
    if pd.api.types.is_numeric_dtype(df[cl]):
        print(f" min {cl}: {df[cl].min()}")
        print(f" max {cl}: {df[cl].max()}")
    print(f" type {cl}: {df[cl].dtype}")

# Генерируем схему
schema = infer_schema_from_df(df)

# Валидируем
validated_df = schema.validate(df)
print("Схема сгенерирована и валидация прошла.")
```

```python
import pandera.pandas as pa

categorical_columns = {
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "default.payment.next.month",
}

delay_shift = {
    "PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6",
}

BILLS = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']

schema = pa.DataFrameSchema(
    {
        cl: pa.Column(
            dtype=df[cl].dtype,
            checks=(
                None
                if cl in BILLS
                else pa.Check.isin(df[cl].dropna().unique().tolist())
                if cl in categorical_columns
                else pa.Check.ge(-3)
                if cl in delay_shift
                else pa.Check.ge(0)
            ),
            nullable=False,
        )
        for cl in df.columns
    },
    index=pa.Index("int64", nullable=False),
    coerce=True,
    strict=False,
)

validated_df = schema.validate(df, lazy=True)
```

```python
# Или сохранить в файл
schema.to_script("schema.py")
```
